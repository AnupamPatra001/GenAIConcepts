```python
prompt='Who is the father of India?'
response= client.chat.completions.create(
    model="openai/gpt-oss-120b",
    messages=[
        {'role':'user', 'content':prompt}
    ],
    temperature=0,
    max_completion_tokens=2048,
    top_p=1,
    reasoning_effort="medium"
    
)

op=response.choices[0].message.content
display_markdown(op)
```

```python
# multi turn conversation
conversation=[]
queries=[
    "I'm planning a trip to Japan. What are the top 3 cities I would visit?",
    "What is the best time of the year to visit those cities?",
    "Can you recommend some local dishes to try there.",
    "How do I get this city to the second city?"
]

for query in queries:
    conversation.append({'role':'user', 'content':query})
    response=client.chat.completions.create(
        model="openai/gpt-oss-120b",
        messages=conversation
    )
    assistant_response=response.choices[0].message.content
    conversation.append({'role':'assistant', 'content':assistant_response})

    display_markdown(f"Human: {query}")
    display_markdown(f"Assistant: {assistant_response}")
```

```python
# ex 4: image analysis
# read and encode image
image_path='Screenshot 2025-03-09 215144.png'
with open(image_path, 'rb') as image_file:
    base64_image=base64.b64encode(image_file.read()).decode('utf-8')
response=client.chat.completions.create(
        model="qwen/qwen3.6-27b",
        messages=[
        {'role':'user', 'content':[
            {
                'type':'text',
                'text':"""
                Analyse the image and complete the following tasks and give output in 500 characters.
                - Describe the main element of the image
                - Identify any visible tect
                - suggest a title for the image
                """
            },
            {
                'type':'image_url',
                'image_url':{
                    'url':f"data:image/jpeg;base64,{base64_image}"
                }
            }
        ]}
        ]
    )

display_markdown(response.choices[0].message.content)
```
